# Preprocessing and Model Formation 

```python
iris = load_iris()
```

```python
x = iris.data
y = iris.target
```

```python
ohe = LabelBinarizer()
y_encoded = ohe.fit_transform(y)
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_scaled, y_encoded, train_size=0.8, random_state=42)
```

```python
from tensorflow.keras.layers import Layer
from tensorflow.keras import backend as K
```

```python
class RBFLayer(Layer):
    def __init__(self, units, gamma, **kwargs):
        super(RBFLayer, self).__init__(**kwargs)
        self.units = units
        self.gamma = K.cast_to_floatx(gamma)

    def build(self, input_shape):
        self.mu = self.add_weight(name='mu', shape=(int(input_shape[1]), self.units),
                                                   initializer = "uniform", trainable = True)
        super(RBFLayer, self).build(input_shape)

    def call(self, inputs):
        diff = K.expand_dims(inputs) - self.mu
        l2 = K.sum(K.pow(diff, 2), axis = 1)
        res = K.exp(-1 * self.gamma * l2)
        return res

    def compute_out_shape(self, input_shape):
        return (input_shape[0], self.units)
```

```python
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense

# Define your EarlyStopping callback
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=5, 
    restore_best_weights=True,
    verbose=1
)

# Define your model
def build_model():
    model = Sequential()
    model.add(Flatten(input_shape=(4,)))  
    model.add(RBFLayer(10, 0.5))         
    model.add(Dense(3, activation="softmax")) 

    model.compile(
        optimizer="adam", 
        metrics=["accuracy"],
        loss="categorical_crossentropy"
    )
    return model


model = build_model()


history = model.fit(
    x_train, y_train,
    epochs=500,
    batch_size=4,
    validation_split=0.3,  
    callbacks=[early_stop],
    verbose=1
)
```

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)
print(f"Test Accuracy: {test_accuracy:.4f}")
print(f"Test Loss: {test_loss:.4f}")
```

# Model Architecture Improvements

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.layers import Layer, Dense, Flatten
from tensorflow.keras.models import Sequential
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from tensorflow.keras import backend as K
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

class RBFLayer(Layer):
    def __init__(self, units, gamma, **kwargs):
        super(RBFLayer, self).__init__(**kwargs)
        self.units = units
        self.gamma = K.cast_to_floatx(gamma)

    def build(self, input_shape):
       
        self.mu = self.add_weight(name='mu',
                                shape=(int(input_shape[1]), self.units),
                                initializer='glorot_uniform',
                                trainable=True)
        super(RBFLayer, self).build(input_shape)

    def call(self, inputs):
        diff = K.expand_dims(inputs) - self.mu
        l2 = K.sum(K.square(diff), axis=1)
        return K.exp(-self.gamma * l2)

    def get_config(self):
        config = super().get_config()
        config.update({
            'units': self.units,
            'gamma': float(self.gamma),
        })
        return config

def build_improved_model():
    model = Sequential([
        Flatten(input_shape=(4,)),
        RBFLayer(16, 0.5),  
        Dense(32, activation='relu'), 
        Dense(3, activation='softmax')
    ])
    
    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy'],
    )
    return model


callbacks = [
    EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True),
    ModelCheckpoint('best_rbf_model.keras', save_best_only=True)
]




```

```python
# Build and train model
model = build_improved_model()
history = model.fit(
    x_train, y_train,
    epochs=500, 
    batch_size=8,  
    validation_split=0.2,
    callbacks=callbacks,
    verbose=1
)
```

```python

# Visualization
plt.figure(figsize=(14,5))
plt.subplot(1,2,1)
plt.plot(history.history['accuracy'], label='Train', color='navy')
plt.plot(history.history['val_accuracy'], label='Validation', color='darkorange')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.title('Training Curves: Accuracy')
plt.subplot(1,2,2)
plt.plot(history.history['loss'], label='Train', color='navy')
plt.plot(history.history['val_loss'], label='Validation', color='darkorange')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.title('Training Curves: Loss')

plt.tight_layout()
plt.show()
```
